```python
import pandas as pd

df_2000 = pd.read_csv('gasolina_2000_.csv')
df_2010 = pd.read_csv('gasolina_2010_.csv', low_memory=False)
```

```python
df_2000.head()
```

```python
df_2000.info()
```

```python
df = pd.concat([df_2000, df_2010], ignore_index=True)
```

```python
df.shape
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.columns
```

```python
terceira_entrada = df['DATA INICIAL'].iloc[2]
print(terceira_entrada)
print(type(terceira_entrada))
```

```python
df['DATA INICIAL'] = pd.to_datetime(df['DATA INICIAL'])
df['DATA FINAL'] = pd.to_datetime(df['DATA FINAL'])
```

```python
df.dtypes
```

```python
df['ANO_MES'] = df['DATA FINAL'].dt.strftime('%Y-%m')
```

```python
df['ANO'] = df['DATA FINAL'].dt.year
```

```python
df[['DATA FINAL', 'ANO_MES', 'ANO']].head()
```

```python
df['PRODUTO'].value_counts()
```

```python
gasolina_comum = df[df['PRODUTO'] == 'GASOLINA COMUM'].copy()
```

```python
gasolina_comum.shape
```

```python
ago_2008 = gasolina_comum[gasolina_comum['ANO_MES'] == '2008-08']
```

```python
preco_medio_ago_2008 = ago_2008['PREÇO MÉDIO REVENDA'].mean()
print(preco_medio_ago_2008)
```

```python
maio_2014_sp = gasolina_comum[
    (gasolina_comum['ANO_MES'] == '2014-05') &
    (gasolina_comum['ESTADO'] == 'SAO PAULO')
]
```

```python
gasolina_comum['ESTADO'].unique()
```

```python
preco_medio_sp_mai2014 = maio_2014_sp['PREÇO MÉDIO REVENDA'].mean()
print(preco_medio_sp_mai2014)
```

```python
acima_de_5 = gasolina_comum[gasolina_comum['PREÇO MÉDIO REVENDA'] > 5.00]
```

```python
estados_acima_de_5 = sorted(acima_de_5['ESTADO'].unique())
print(estados_acima_de_5)
```

```python
primeira_data_acima_5 = acima_de_5.groupby('ESTADO')['DATA FINAL'].min().sort_values()
print(primeira_data_acima_5)
```

```python
sul_2012 = gasolina_comum[
    (gasolina_comum['REGIÃO'] == 'SUL') &
    (gasolina_comum['ANO'] == 2012)
]
```

```python
media_sul_2012 = sul_2012['PREÇO MÉDIO REVENDA'].mean()
print(media_sul_2012)
```

```python
sul_2012.groupby('ESTADO')['PREÇO MÉDIO REVENDA'].mean()
```

```python
rj = gasolina_comum[gasolina_comum['ESTADO'] == 'RIO DE JANEIRO']
```

```python
preco_medio_rj_por_ano = rj.groupby('ANO')['PREÇO MÉDIO REVENDA'].mean()
print(preco_medio_rj_por_ano)
```

```python
variacao_percentual = preco_medio_rj_por_ano.pct_change() * 100
```

```python
tabela_variacao_rj = pd.DataFrame({
    'preco_medio': preco_medio_rj_por_ano,
    'variacao_%_ano_a_ano': variacao_percentual
})
print(tabela_variacao_rj)
```
